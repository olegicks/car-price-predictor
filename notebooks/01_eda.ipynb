```python
print("Rows:", len(df))

print("Price <= 1000:", (df["price"] <= 1000).sum())
print("Price >= 1,000,000:", (df["price"] >= 1_000_000).sum())

print("Mileage == 999999:", (df["mileage"] == 999999).sum())
print("Mileage > 500000:", (df["mileage"] > 500_000).sum())

print("Exact duplicates:", df.duplicated().sum())
```

```python
df[df["price"] >= 1_000_000][
    ["manufacturer", "model", "year", "mileage", "price"]
].sort_values("price", ascending=False).tail(30)
```

```python
df_clean = df.copy()

# Remove exact duplicates
df_clean = df_clean.drop_duplicates()

# Remove clearly invalid prices
df_clean = df_clean[
    (df_clean["price"] > 1000) &
    (df_clean["price"] < 1_000_000)
]

# Replace clearly invalid mileage with missing values
df_clean.loc[df_clean["mileage"] > 500_000, "mileage"] = np.nan

print("Original rows:", len(df))
print("Cleaned rows:", len(df_clean))
print("Rows removed:", len(df) - len(df_clean))

print("\nRemaining missing values:")
print(df_clean.isnull().sum())
```
